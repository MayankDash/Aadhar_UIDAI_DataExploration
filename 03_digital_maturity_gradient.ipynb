```python
from sklearn.preprocessing import StandardScaler
import pandas as pd
```

```python
digital_maturity_df = pd.read_csv("data/digital_maturity_features.csv")
```

```python
digital_maturity_df.head()
```

```python
feature_cols = [
    "mean_stress_index",
    "mean_update_dependency",
    "stress_volatility",
    "dependency_volatility",
    "pct_active_months",
    "activity_growth_slope"
]
```

```python
# Checking for any NaN values
digital_maturity_df[feature_cols].isna().any().any()
```

```python
X = digital_maturity_df[feature_cols]


### Standardizing the dataset
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

### KMeans

```python
from sklearn.cluster import KMeans
```

```python
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
digital_maturity_df["maturity_cluster"] = kmeans.fit_predict(X_scaled)
```

```python
### Understanding and labeling the clusters

cluster_profiles = (
    digital_maturity_df
    .groupby("maturity_cluster")[feature_cols]
    .mean()
)

cluster_profiles
```

```python
cluster_name_map = {
    0: "High Activity – Unstable / Transitioning",
    1: "Emerging Growth",
    2: "Dormant / Low Coverage",
    3: "High Stress / Legacy Heavy"
}
```

```python
mean_mothly_activity = pd.read_csv("data/mean_monthly_activity.csv")
```

```python
digital_maturity_df = (
    digital_maturity_df
    .merge(
        mean_mothly_activity,
        on=["state", "district"],
        how="left"
    )
)
```
